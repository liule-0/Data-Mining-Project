# 📊 人口结构分析 — 数据加载与探索性分析

**目标：** 加载重庆市统计局各区县人口数据，进行初步探索和可视化

```python
import os
import sys
sys.path.insert(0, os.path.join(os.getcwd(), '..', '..'))

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.rcParams['font.sans-serif'] = ['Microsoft YaHei']
plt.rcParams['axes.unicode_minus'] = False

from src.utils.config_loader import get_config
config = get_config()
print('配置加载完成:', config['project']['name'])
```

## 1. 加载数据

```python
# 加载人口数据
from src.data.data_loader import load_population_data

df_pop = load_population_data()
print(f'数据维度: {df_pop.shape}')
df_pop.head()
```

```python
# 数据基本信息
df_pop.info()
```

```python
# 描述统计
df_pop.describe()
```

## 2. 数据清洗

```python
from src.data.data_preprocessor import clean_population_data, compute_aging_rate, region_classify

df_clean = clean_population_data(df_pop)
df_clean = compute_aging_rate(df_clean)
df_clean['区域'] = df_clean['区县'].apply(region_classify)
print('清洗完成')
df_clean.head()
```

## 3. 探索性分析

```python
# 各区县老龄化率分布
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# 箱线图
sns.boxplot(data=df_clean, x='区域', y='老龄化率', ax=axes[0])
axes[0].set_title('各区域老龄化率分布')

# 直方图
sns.histplot(data=df_clean, x='老龄化率', hue='区域', kde=True, ax=axes[1])
axes[1].set_title('老龄化率分布直方图')

plt.tight_layout()
plt.show()
```

```python
# 按年份统计
df_clean.groupby('年份')['老龄化率'].agg(['mean', 'std', 'min', 'max']).round(4)
```
